```python
import numpy as np
```

```python
x=np.arange(24).reshape(2,3,4)
print("Shape:",x.shape)
print("dimensions:",x.ndim)
print("number of elements:",x.size)
print("data type:",x.dtype)
print("byter per element:",x.itemsize)
print("total bytes:",x.nbytes)
v=np.array([10,20,30,40,50])
print(v[0],v[-1])
print(v[1:4])
print(v[::2])
print(v[::-1])
```
