```python
import numpy as np
```

```python
rng=np.random.default_rng(seed=42)
readings=rng.random((9,9))
print(readings)
```

```python
print("The mean:",np.mean(readings))
```

```python
print("The standard:",np.std(readings))
```

```python
print("The median:",np.median(readings))
```

```python
mean=np.mean(readings)
flags=readings[mean>readings]
anomalies=(mean-flags)
print(flags+anomalies)
```
