```python
import numpy as np
```

```python
k1=np.arange(0,5, dtype=np.uint8)
print(k1.dtype, k1)
```

```python
k2=k1+np.array([5,6,7,8,9], dtype=np.int8)
print(k2.dtype, k2)
```

```python
k3=k1+1.5
print(k3.dtype, k3)
```

```python
m=np.array([20,-5,30,40])
print(m.shape)
print(m<[15,16,35,36])
print(m<25)
print(m[m<25])
```

```python
a = np.array([[-2.5, 3.1, 7], [10, 11, 12]])
print(a)
print("Mean=",a.mean())
```

```python
print("Sum=",a.sum())
```

```python
a=np.array([[-2.5,3.1,7],[10,11,12]])
np.square(a)
```

```python
for func in (np.abs, np.sqrt, np.exp, np.log, np.sign, np.ceil, np.modf, np.isnan, np.cos):
    print("\n", func.__name__)
    print(func(a))
```
