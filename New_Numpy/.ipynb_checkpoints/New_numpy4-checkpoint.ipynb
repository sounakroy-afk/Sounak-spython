## NumPy Exercises

1. Create a random 4×5 array of values between 0 and 1. Print its shape, data type, and smallest and largest values.
2. Generate a 3×6 array from a standard normal distribution. Count how many values are positive and how many are negative.
3. Use `rng.integers` to make a 5×4 array of random integers from 12 up to (but not including) 30. Replace every value below 20 with `-1`.
4. Generate 12 random integers in the range 1–50. Sort a copy of them and compare it with the original, keeping the original order intact.
5. Create a permutation of the numbers 0 through 14. Use it to split the numbers into two groups of different sizes, with no repeats.
6. Shuffle a list of weekday names using a permutation of indices. Print the shuffled names alongside their original indices.
7. Initialize two random generators with the same seed. Generate a 2×3 array from each. What do you observe, and why might a fixed seed be useful?
8. Write a function suitable for `np.fromfunction` that returns `row + 2 * column`. Use it to build a 4×5 array, then explain how the values change across rows and columns.
9. Use `np.fromfunction` to create a 3×4 array whose values are `row * column`. Identify which positions contain zero and explain why.
10. Create a 3×4×3 array with `np.fromfunction` where each value is `100 * first_index + 10 * second_index + third_index`. Use a few entries to verify that you understand how the three indices map to values.
11. Use `np.fromfunction` to build a 5×5 array where each entry is the square of its row index minus its column index. Then extract the main diagonal.
12. Make an array with `np.arange` from 1 to 20, stepping by 3. Predict the final value before printing the array.
13. Create an `np.arange` array from 0.5 to 3.0 with a step of 0.25. Check its data type and explain whether the endpoint appears.
14. Use `np.arange` to generate the even numbers from 2 up to 40. Reshape them into a 2×10 array and calculate the sum of each row.
15. Make an integer array with `np.arange`, then convert it to floating-point values. Compare the data types before and after conversion.
16. Use a floating-point `np.arange` sequence to create values from 0 to 1 in steps of 0.1. Check twhether the values match exact decimal tenths, and explain any surprising output.
17. Build a 4×4 grid of row and column indices using `np.fromfunction`. Use the grid to select entries where the row index is greater than the column index.
18. Generate a random 6×6 integer array. Use a `np.fromfunction`-created mask or index grid to replace values on the main diagonal with zero.
19. Create a random permutation of 0 through 19, then use it to reorder a separate array containing 20 letters or numbers. Verify that every original item appears exactly once.
20. Combine the notebook’s ideas: create a 4×5 array with `np.fromfunction`, add random integers to it, and then use `np.arange` to produce row or column labels. Print the result with the labels and describe how each part was constructed.

Several prompts have multiple valid solutions. Try to solve them using the NumPy tools shown in this notebook.


```python
rng=np.random.default_rng()
array1=rng.random((4,5))
print(array1)
print("The shape:",array1.shape, "\nThe data type:",array1.dtype)
print("The minimum:",np.min(array1),"\nThe maximum:",np.max(array1))
```

```python
array1=rng.standard_normal((3,6))
array2=array1>0
array3=array1<0
print("The original values are:",array1)
print("The values which are negative:",array2)
print("The values which are positive:",array3)
```
