# NumPy 3: the next tools to master

Your first notebook covers creating arrays, inspecting shape and dtype, sorting, concatenating, reshaping, slicing, boolean masks, and `np.nonzero`. This lesson builds on those skills. Run the cells from top to bottom; change the values and predict the output as you go.

```python
import numpy as np

np.set_printoptions(precision=3, suppress=True)
```

## 1. Copying, changing dtype, and initializing arrays

Assignment creates another reference to the same array. Use `.copy()` when you need independent data. `astype` converts element types; `full` and `eye` create useful structured arrays.

```python
original = np.array([1, 2, 3])
reference = original
independent = original.copy()
reference[0] = 99
print('original changed through reference:', original)
print('independent copy:', independent)

values = np.array([1.7, 2.2, 3.9])
print('as integers (decimal part is dropped):', values.astype(np.int32))
print('filled array:', np.full((2, 3), 7))
print('identity matrix:', np.eye(3, dtype=int))
```

## 2. Combine and split arrays

`stack` adds a new axis. `vstack` and `hstack` join along existing row and column directions. `array_split` can produce sections that are not all the same size.

```python
left = np.array([1, 2, 3])
right = np.array([4, 5, 6])
print('stack, rows:', np.stack((left, right), axis=0))
print('stack, columns:', np.stack((left, right), axis=1))
print('vertical:', np.vstack((left, right)))
print('horizontal:', np.hstack((left, right)))

parts = np.array_split(np.arange(10), 3)
print('split into 3 parts:', parts)
```

## 3. Transpose and change axes

A transpose swaps axes. For a 2D matrix, `.T` is the common shorthand. `swapaxes` names the two axes to exchange, and `moveaxis` relocates an axis.

```python
cube = np.arange(24).reshape(2, 3, 4)
print('cube shape:', cube.shape)
print('transpose shape:', cube.transpose(1, 0, 2).shape)
print('swap first and last axes:', np.swapaxes(cube, 0, 2).shape)
print('move axis 0 to the end:', np.moveaxis(cube, 0, -1).shape)

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print('matrix transpose:\n', matrix.T)
```

## 4. Broadcasting and elementwise operations

Broadcasting lets compatible shapes work together without manually repeating data. Dimensions are compatible when they are equal or one of them is 1.

```python
measurements = np.array([[10.0, 20.0, 30.0],
                         [12.0, 18.0, 33.0]])
column_offsets = np.array([1.0, 2.0, 3.0])
print('add one offset per column:\n', measurements + column_offsets)
print('multiply every value by 2:', measurements * 2)
print('square root:', np.sqrt(measurements))
print('sine (input is in radians):', np.sin(measurements))
```

## 5. Conditional selection, uniqueness, and counts

`where` selects values element by element. `unique` can return counts and inverse indices. `count_nonzero` is handy for counting true conditions.

```python
scores = np.array([42, 78, 91, 59, 78, 100])
labels = np.where(scores >= 60, 'pass', 'retry')
print('labels:', labels)
unique_scores, counts = np.unique(scores, return_counts=True)
print('unique scores:', unique_scores)
print('frequency:', counts)
print('number passing:', np.count_nonzero(scores >= 60))
```

## 6. Reductions and axis-aware summaries

Reduction functions combine values. With a 2D array, `axis=0` works down the rows (one result per column), while `axis=1` works across columns (one result per row). `keepdims=True` preserves the reduced dimensions for broadcasting.

```python
data = np.array([[2, 4, 6],
                  [1, 3, 8],
                  [5, 7, 9]])
print('sum:', np.sum(data))
print('column means:', np.mean(data, axis=0))
print('row maxima:', np.max(data, axis=1))
print('column standard deviations:', np.std(data, axis=0))
print('min/max:', np.min(data), np.max(data))
print('minimum positions by column:', np.argmin(data, axis=0))
print('maximum positions by row:', np.argmax(data, axis=1))
column_means = np.mean(data, axis=0, keepdims=True)
print('center columns:\n', data - column_means)
```

## 7. Percentiles and clipping

`percentile` describes a position in sorted data. `clip` limits values to a specified range.

```python
temps = np.array([18, 21, 19, 25, 31, 27, 22])
print('median:', np.median(temps))
print('25th, 50th, 75th percentiles:', np.percentile(temps, [25, 50, 75]))
print('limited to 20..28:', np.clip(temps, 20, 28))
```

## 8. Random numbers with a generator

Create a `default_rng` generator. A seed makes results repeatable. Use `integers` for integer samples and `normal` for normally distributed samples.

```python
rng = np.random.default_rng(seed=42)
print('integers from 0 through 9:', rng.integers(0, 10, size=5))
print('uniform samples:', rng.random(size=(2, 3)))
print('normal samples:', rng.normal(loc=0, scale=1, size=4))
```

## 9. Missing values and NaN-safe functions

`NaN` is a floating-point marker for missing or invalid numeric data. Normal means propagate NaN; `nanmean` skips it. Use `isnan` to find missing positions.

```python
readings = np.array([20.0, np.nan, 23.0, 21.0])
print('missing mask:', np.isnan(readings))
print('ordinary mean:', np.mean(readings))
print('mean skipping NaN:', np.nanmean(readings))
filled = np.where(np.isnan(readings), np.nanmean(readings), readings)
print('missing value filled with mean:', filled)
```

## 10. Linear algebra

`@` and `matmul` perform matrix multiplication; `dot` is also commonly used for dot products. `linalg.solve` solves a square system `A @ x = b`. `linalg.norm` measures vector length.

```python
A = np.array([[3.0, 1.0], [1.0, 2.0]])
b = np.array([9.0, 8.0])
x = np.linalg.solve(A, b)
print('solution x:', x)
print('check A @ x:', A @ x)

u = np.array([1.0, 2.0, 2.0])
v = np.array([2.0, 0.0, 1.0])
print('dot product:', np.dot(u, v))
print('vector length:', np.linalg.norm(u))
print('determinant:', np.linalg.det(A))
```

## 11. Put it together: normalize columns

This example combines axis reductions, broadcasting, and safe division. Each column is shifted to mean zero and scaled to standard deviation one.

```python
samples = np.array([[2.0, 10.0, 5.0],
                      [4.0, 12.0, 5.0],
                      [6.0, 14.0, 5.0]])
means = samples.mean(axis=0, keepdims=True)
stds = samples.std(axis=0, keepdims=True)
# The third column is constant, so replace its zero standard deviation with 1.
stds_safe = np.where(stds == 0, 1, stds)
normalized = (samples - means) / stds_safe
print('means:', means)
print('standard deviations:', stds)
print('normalized:\n', normalized)
```
