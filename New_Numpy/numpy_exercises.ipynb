# NumPy Exercises

Work through these questions using NumPy. Unless a question asks for a written explanation, write and run code to produce the requested result. For random exercises, use `np.random.default_rng()`; use a seed when reproducible output is requested.

## Arrays and creation

1. Create a 3 by 4 array containing the integers 1 through 12. Print the array, its number of dimensions, shape, size, and data type.
2. For the array from Question 1, verify that its size equals the product of its shape dimensions. Use `math.prod`.
3. Create a 2 by 3 array of zeros, a 2 by 3 array of ones, and a 3 by 3 identity matrix. Set the identity matrix's dtype to an integer type.
4. Create an array of six uninitialized floating-point values with `np.empty`. Print its shape and dtype. Why should you not expect particular values in it?
5. Use `np.arange` to create the even numbers from 2 up to (but not including) 20. Then use `np.linspace` to create seven evenly spaced values from -1 through 1.
6. Create arrays using `np.arange(1, 5)` and `np.arange(1.0, 5.0)`. Print and compare their dtypes. Convert the floating-point array to `int32` with `astype`.
7. Create an array of shape `(2, 3)` filled with `-4` using `np.full`. Create a 4 by 4 identity matrix using `np.eye`.
8. Use `np.fromfunction` and a function of three coordinate arguments to create an array of shape `(2, 3, 4)` whose values are `x + y * z`. Print the result.

## Copying, reshaping, and axes

9. Assign a NumPy array to a second variable, change an element through the second variable, and show that the original variable sees the change.
10. Repeat Question 9 using `.copy()`. Change the copied array and show that the original is unchanged.
11. Reshape `np.arange(12)` to `(3, 4)` using both the array's `.reshape` method and `np.reshape`. Print both shapes.
12. Reshape `np.arange(6)` to `(2, 3)` using `order='C'` and `order='F'`. Print and explain the difference in how values are placed.
13. Turn `np.arange(6)` into a row vector and a column vector using `np.newaxis`. Print both shapes.
14. Use `np.expand_dims` to add an axis to a one-dimensional array. Try both axis 0 and axis 1 and report the resulting shapes.
15. Create an array of shape `(2, 3, 4)`. Print the shapes produced by `.transpose(1, 0, 2)`, `np.swapaxes(array, 0, 2)`, and `np.moveaxis(array, 0, -1)`.
16. Create a 2 by 3 matrix, print its transpose using `.T`, and verify that transposing twice returns the original values.
17. Flatten a 3 by 4 array with `ravel`. Print the flattened values and explain their order.

## Indexing, filtering, sorting, joining, and splitting

18. For `np.array([10, 20, 30, 40, 50, 60])`, print the element at index 2 and the slices `[:3]`, `[2:]`, and `[1:5:2]`.
19. Create a 3 by 4 array and select one element, its second row, its third column, and a 2 by 2 rectangular section using indexing and slicing.
20. Given a 3 by 4 array containing 1 through 12, select all values less than 6 using a Boolean condition.
21. Using the same array, select values at most 5 and values divisible by 2. Create and print each Boolean mask as well as the selected values.
22. Use `np.nonzero` to find row and column coordinates where a 3 by 4 array is greater than 8. Pair and print the coordinates. What does `np.nonzero` return if no elements match?
23. Sort an array containing repeated values with `np.sort`. Use `np.argsort` to get the indices that would sort it, then use those indices to reorder the original array.
24. Create two length-4 arrays. Join them with `np.concatenate`, `np.stack` along axis 0, `np.stack` along axis 1, `np.vstack`, and `np.hstack`. Print the shape of every result.
25. Concatenate two 2D arrays along axis 0 and then along axis 1. Choose compatible shapes and explain which dimensions must match for each operation.
26. Split `np.arange(17)` into four sections using `np.array_split`. Print each section and its shape. Why can the sections have different sizes?

## Broadcasting and elementwise math

27. Create a `(3, 4)` measurements array and a length-4 offsets array. Add and subtract the offsets so each column is adjusted. Explain why the shapes broadcast.
28. Create an array with shape `(1, 1, 5)` and add a length-5 array. Print the result and its shape.
29. For a 2 by 3 array, separately add `[[100], [200]]`, `[10, 20, 30]`, and the scalar `1000`. Print each result and describe the compatible dimensions.
30. Create Celsius temperatures and convert them to Fahrenheit using `F = C * 9/5 + 32`. Apply `np.sqrt` to nonnegative values and `np.sin` and `np.cos` to angles in radians.
31. Demonstrate elementwise addition with both `np.add(a, b)` and `a + b`. Use `np.greater(a, b)` to compare two arrays element by element.
32. Given two arrays with positive and negative values, print the results of `np.maximum(a, b)` and `np.copysign(a, b)`. Explain what differs between them.
33. Create an array containing positive, negative, and zero values. Apply `np.square`, `np.abs`, `np.sign`, `np.ceil`, `np.exp`, and `np.log` to suitable inputs. Use `np.modf` to separate fractional and integer parts.

## Random numbers and summaries

34. Create two random generators with the same seed. Generate ten integers from 1 through 6 from each and verify that the sequences match. Remember that the upper bound of `integers` is exclusive.
35. Using a seeded generator, create a `(3, 4)` array of uniform random values and another `(3, 4)` array of standard normal values. Print both.
36. Generate a `(4, 3)` integer array from 10 up to but not including 20. Report its minimum and maximum.
37. Use `permutation` to randomly reorder the integers 0 through 9. Verify that every original value appears exactly once. Does the input sequence get changed in place?
38. Create a numeric 3 by 4 array. Calculate the overall sum, row sums, and column sums using appropriate `axis` values. State what `axis=0` and `axis=1` mean here.
39. Calculate the overall mean, row means, column means, and column standard deviations of a 2D array. Use `keepdims=True` for the column means, subtract them from the data, and confirm the centered column means are close to zero.
40. For a 2D array, find its minimum and maximum values, the index of each column's minimum using `argmin`, and the index of each row's maximum using `argmax`.
41. For a dataset of your choice, calculate its median and 25th, 50th, and 90th percentiles. Use `np.clip` to constrain its values to a range and print the original and clipped data.
42. Label scores as `"pass"` or `"retry"` with `np.where` using 60 as the threshold. Use `np.unique(..., return_counts=True)` to report score frequencies and `np.count_nonzero` to count passing scores.
43. Create a 3D array and calculate sums over `axis=0`, `axis=1`, and `axis=(0, 2)`. Print the shape of each result and explain which dimensions were reduced.
44. Create 1,000 normally distributed samples with mean 5 and standard deviation 2. Report the sample mean and standard deviation. How do you expect they compare with 5 and 2?

## Missing data and linear algebra

45. Create a floating-point array containing two `np.nan` values. Locate them with `np.isnan`, compare `np.mean` with `np.nanmean`, and fill missing entries with the NaN-safe mean using `np.where`.
46. Use `np.isfinite(...).all()` to check that an array contains only finite values. Include a `NaN` or infinity and show how the check changes.
47. Calculate the dot product of two length-3 vectors with `np.dot`. Calculate the length of one vector with `np.linalg.norm`.
48. Multiply two compatible matrices using `@` and `np.matmul`. Verify the output shape, and calculate the determinant of a square matrix using `np.linalg.det`.
49. Solve a 3 by 3 linear system `A @ x = b` using `np.linalg.solve`. Verify the solution with `np.allclose(A @ x, b)`.

## Integrated challenge

50. Create a 5 by 4 measurement dataset with one constant-valued column and one `np.nan` entry. Find the missing entry and replace it with the mean of its column using `np.nanmean` and `np.where`. Calculate column means and standard deviations with axis reductions and `keepdims=True`; normalize each column to mean zero and standard deviation one, safely handling the constant column by substituting 1 for a zero standard deviation. Confirm that all output values are finite with `np.isfinite(...).all()`, check the nonconstant columns' means are close to zero, and count normalized values above zero with `np.count_nonzero`.
