```python
import pandas as pd
```

```python
print(pd.__version__)
```

```python
data = [100,102,104]
series = pd.Series(data)
print(series)
```

```python
data = [100.1,102.2,104.3]
series = pd.Series(data)
print(series)
```

```python
data = ["A","B","C"]
series = pd.Series(data)
print(series)
```

```python
data = [True, False, True]
series = pd.Series(data)
print(series)
```

```python
data = [100,102,104]
series = pd.Series(data, index=["a","b","c"])
print(series)
print(series.loc["a"])
```

```python
data = [100,102,104]
series = pd.Series(data, index=["a","b","c"])
print(series)
series.loc["a"]=200
print(series)
print(series.loc["a"])
```

```python
data = [100,102,104]
series = pd.Series(data, index=["a","b","c"])
print(series.iloc[0])
```

```python
data = [100,102,104,200,202]
series = pd.Series(data, index=["a","b","c","d","e"])
print(series[series >= 200])
```

```python
calories={"Day 1":1750, "Day 2":2100, "Day 3":1700}
series=pd.Series(calories)
series.loc["Day 3"] += 500
print(series)
print(series[series >= 2000])
```
